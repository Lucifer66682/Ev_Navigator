```python
from transformers import AutoTokenizer, AutoModel
import torch
import pandas as pd
import json

# -----------------------------
# 1. Load MuRIL
# -----------------------------
MODEL_NAME = "google/muril-base-cased"

print("Loading MuRIL model...")

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModel.from_pretrained(MODEL_NAME)

print("MuRIL loaded successfully!\n")


# -----------------------------
# 2. Load sample complaint data
# -----------------------------
df = pd.read_csv("sample_texts_final_merged.csv")

print("Number of samples:", len(df))
print("Columns:", df.columns.tolist())
print()


# -----------------------------
# 3. Process each complaint
# -----------------------------
for index, row in df.iterrows():

    text = row["text"]

    print(f"Sample {row['id']}")
    print("Text:", text)

    # Convert text into tokens
    inputs = tokenizer(
        text,
        return_tensors="pt",
        truncation=True,
        padding=True
    )

    # Generate embedding
    with torch.no_grad():
        outputs = model(**inputs)

    # CLS token embedding
    embedding = outputs.last_hidden_state[:, 0, :]

    print("Embedding shape:", embedding.shape)
    print("First 10 values:", embedding[0][:10].tolist())
    print("-" * 60)


# -----------------------------
# 4. Load crisis keywords
# -----------------------------
with open("crisis_keywords_final_merged.json", "r", encoding="utf-8") as file:
    crisis_keywords = json.load(file)

print("\nCrisis keyword file loaded successfully.")

print(
    "Languages:",
    crisis_keywords["metadata"]["languages"]
)
```
